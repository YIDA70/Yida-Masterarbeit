```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# 数据文件
src = Path(r"D:\毕设数据\20_export_pulse\20_export_pulse\METABatt_Sony_Murata_18650VTC6_007\METABatt_Sony_Murata_18650VTC6_007_pulse_BM12_90.3SOH.parquet")

# 输出图片
out = Path(r"D:\毕设数据\code\pulse_start_zoom_12_21_current_voltage_close.png")

# 选择一个 pulse
pulse_id = "12_21"

# pulse 起点前后窗口，单位 s
pre_s = 2
post_s = 6

# 读取数据
df = pd.read_parquet(src).sort_values("Time").reset_index(drop=True)

# 找到该 pulse 的开始时间
start_time = df.loc[df["ID"].eq(pulse_id), "Time"].min()

# 截取 pulse 开始附近的数据
window = df[
    (df["Time"] >= start_time - pd.Timedelta(seconds=pre_s)) &
    (df["Time"] <= start_time + pd.Timedelta(seconds=post_s))
].copy()

# 转成相对 pulse 起点的时间
window["t_rel_s"] = (window["Time"] - start_time).dt.total_seconds()

# 画 Current + Voltage 双 y 轴
fig, ax_i = plt.subplots(figsize=(9.5, 5.2), dpi=180)
ax_v = ax_i.twinx()

for seg_id, g in window.groupby("ID", sort=False):
    alpha = 1.0 if seg_id == pulse_id else 0.60
    lw = 1.6 if seg_id == pulse_id else 1.0

    ax_i.plot(
        g["t_rel_s"], g["Current"],
        color="#006BA4", marker="o", ms=2.5, lw=lw, alpha=alpha
    )

    ax_v.plot(
        g["t_rel_s"], g["Voltage"],
        color="#D55E00", marker="o", ms=2.5, lw=lw, alpha=alpha
    )

# pulse 起点标记
ax_i.axvline(0, color="0.2", lw=1.0, ls="--")
ax_i.annotate(
    "pulse start",
    xy=(0, window.loc[window["ID"].eq(pulse_id), "Current"].iloc[0]),
    xytext=(0.25, 1.15),
    arrowprops=dict(arrowstyle="->", color="0.25", lw=0.8),
    fontsize=9,
    color="0.25"
)

ax_i.set_title(f"ID {pulse_id} pulse start close-up")
ax_i.set_xlabel("Time relative to pulse start (s)")
ax_i.set_ylabel("Current (A)", color="#006BA4")
ax_v.set_ylabel("Voltage (V)", color="#D55E00")

ax_i.tick_params(axis="y", labelcolor="#006BA4")
ax_v.tick_params(axis="y", labelcolor="#D55E00")

ax_i.grid(True, which="major", alpha=0.25)
ax_i.set_xlim(-pre_s, post_s)

fig.text(
    0.5, 0.012,
    f"{pulse_id}: {start_time}",
    ha="center",
    fontsize=8,
    color="0.35"
)

fig.tight_layout(rect=(0, 0.04, 1, 1))
fig.savefig(out, bbox_inches="tight")

plt.show()
```
